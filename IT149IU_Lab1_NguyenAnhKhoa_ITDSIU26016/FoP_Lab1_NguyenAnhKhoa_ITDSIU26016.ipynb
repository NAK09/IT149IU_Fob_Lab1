## Exercise 2 — Range checking and squares


```python
numbers = [2, 3.5, 10, 0, 8]
for i, number in enumerate(numbers, 1):
    if 1 <= number <= 10:
        print(f'number {i}: Square = {number ** 2}')
    else:
        print(f'number {i}: Error: Number is out of range [1, 10]')
```

## Exercise 4 — Packing eggs


```python
eggs, per_box = 53, 12
boxes = eggs // per_box + (1 if eggs % per_box else 0)
last_box = eggs % per_box if eggs % per_box else per_box
empty = per_box - eggs % per_box if eggs % per_box else 0
print(f'Boxes needed: {boxes}')
print(f'Eggs in last box: {last_box}')
print(f'Eggs needed to fill last box: {empty}')
```

## Exercise 5 — Loops and DataFrame filtering


```python
for number in range(1, 11):
    print(f'{number} is {"even" if number % 2 == 0 else "odd"}')
df = pd.DataFrame({'Data': [10, 20, 30, 40, 50, 60]})
print('Even index values:', df[df.index % 2 == 0]['Data'].tolist())
print('Odd index values:', df[df.index % 2 != 0]['Data'].tolist())
```

## Exercise 6 — Bacteria growth and plotting


```python
hours = list(range(0, 16, 5))
bacteria = [200 * (2 ** h) for h in hours]
print(list(zip(hours, bacteria)))
plt.plot(hours, bacteria, marker='o')
plt.title('Bacteria Growth Curve')
plt.close()
```

## Exercise 8 — Functions and name scores


```python
def name_score(name):
    return sum(ord(c) for c in name)
names = ['Ana', 'Bao', 'Minh']
for name in names:
    print(f'{name} score: {name_score(name)}')
print(max(names, key=name_score), 'goes first!')
```

## Exercise 10 — Seconds to clock time


```python
total_seconds = 3671
hours = total_seconds // 3600
remainder = total_seconds % 3600
minutes, seconds = remainder // 60, remainder % 60
print(f'{hours:02d}:{minutes:02d}:{seconds:02d}')
print('Verification:', hours * 3600 + minutes * 60 + seconds == total_seconds)
```

## Exercise 11 — Wage changes and plotting


```python
initial, rate, years = 20.0, 0.03, 4
final_wage = initial * ((1 + rate) ** years)
current = initial
for review in ['good', 'bad', 'good']:
    if review == 'good': current *= 1.03
    elif review == 'bad': current *= 0.97
print(f'Compound wage: {final_wage:.2f} USD')
print(f'Review sequence wage: {current:.2f} USD')
```

## Exercise 12 — Sorting runner times


```python
times = {'Runner 1': 12.4, 'Runner 2': 11.8, 'Runner 3': 12.1}
ordered = sorted(times.items(), key=lambda item: item[1])
print('Times in ascending order:', [time for _, time in ordered])
print(ordered[0][0], 'won!')
```

## Exercise 13 — BMI classification


```python
weight, height = 68, 1.70
bmi = weight / height ** 2
category = 'Underweight' if bmi < 18.5 else 'Normal' if bmi < 25 else 'Overweight' if bmi < 30 else 'Obese'
print(f'BMI = {bmi:.1f} -> {category}')
```

## Exercise 14 — VND denomination breakdown


```python
amount = 1_250_000
for note in [500_000, 200_000, 100_000, 50_000]:
    count, amount = divmod(amount, note)
    print(f'{note:,} x {count}')
```

## Exercise 15 — Digits, reversal and palindrome


```python
n = 121
hundreds, tens, units = n // 100, (n // 10) % 10, n % 10
reversed_num = units * 100 + tens * 10 + hundreds
print('Digits:', hundreds, tens, units)
print('Sum of digits:', hundreds + tens + units)
print('Reversed number:', reversed_num)
print('Palindrome:', n == reversed_num)
```
